# Ćwiczenie: Wpływ inhibicji produktem na konwersję substratu w reaktorze okresowym

**Biokataliza przemysłowa — Python**

**Cel:** Określenie wpływu inhibicji produktem na czas potrzebny do uzyskania 90% i 95% konwersji substratu.

Rozważamy reakcję:

$$S \longrightarrow P$$

Szybkość reakcji opisuje równanie Michaelisa–Menten z **kompetycyjną inhibicją produktem**:

$$v=\frac{v_{\max}S}{K_m(1+P/K_i)+S}$$

Dla reaktora okresowego (stała objętość, stechiometria 1:1):

$$\frac{dS}{dt}=-v, \qquad \frac{dP}{dt}=v$$

Przy $P_0=0$ obowiązuje bilans:

$$P=S_0-S$$

Konwersja substratu:

$$X(t)=\frac{S_0-S(t)}{S_0}\cdot100\%$$

**Dane do obliczeń:**

- $S_0=100$ mmol/L, $P_0=0$ mmol/L
- $v_{\max}=10$ mmol/(L·min)
- $K_m=20$ mmol/L
- Warianty: brak inhibicji oraz $K_i=100$, $30$, $10$ mmol/L

*Model i parametry są dydaktyczne.*

## 1. Przykład — symulacja dla $K_i=30$ mmol/L

Uruchom poniższy kod. Zwróć uwagę, że funkcja `model(y, t)` zwraca pochodne stężeń, a `odeint` oblicza zmiany stężeń w czasie.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint

vmax = 10
Km = 20
Ki = 30

# Model kinetyczny
def model(y, t):
    S, P = y

    v = vmax * S / (Km * (1 + P/Ki) + S)

    return [-v, v]

# Warunki początkowe
S0 = 100
P0 = 0

czas = np.linspace(0, 60, 301)

# Rozwiązanie równań
wynik = odeint(model, [S0, P0], czas)

S = wynik[:, 0]
P = wynik[:, 1]

# Konwersja substratu
X = (S0 - S) / S0 * 100

# Wykres
plt.plot(czas, X)
plt.xlabel("Czas [min]")
plt.ylabel("Konwersja [%]")
plt.grid()
plt.show()
```
